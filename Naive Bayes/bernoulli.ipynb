```python
from sklearn.datasets import fetch_20newsgroups
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB, BernoulliNB
from sklearn.metrics import accuracy_score, f1_score, classification_report


categories = ['alt.atheism', 'comp.graphics', 'sci.space', 'talk.religion.misc']
print("Downloading/Loading 20 Newsgroups data...")
train_data = fetch_20newsgroups(subset='train', categories=categories, remove=('headers', 'footers', 'quotes'))
test_data = fetch_20newsgroups(subset='test', categories=categories, remove=('headers', 'footers', 'quotes'))

y_train = train_data.target
y_test = test_data.target


vectorizer_mnb = CountVectorizer(max_features=10000, stop_words='english')
X_train_mnb = vectorizer_mnb.fit_transform(train_data.data)
X_test_mnb = vectorizer_mnb.transform(test_data.data)


vectorizer_bnb = CountVectorizer(max_features=10000, stop_words='english', binary=True)
X_train_bnb = vectorizer_bnb.fit_transform(train_data.data)
X_test_bnb = vectorizer_bnb.transform(test_data.data)


mnb = MultinomialNB(alpha=0.1) 
mnb.fit(X_train_mnb, y_train)

bnb = BernoulliNB(alpha=0.1)
bnb.fit(X_train_bnb, y_train)


preds_mnb = mnb.predict(X_test_mnb)
preds_bnb = bnb.predict(X_test_bnb)

print("=== Multinomial Naive Bayes ===")
print(f"Accuracy: {accuracy_score(y_test, preds_mnb):.4f}")
print(f"F1-Score (Weighted): {f1_score(y_test, preds_mnb, average='weighted'):.4f}\n")

print("=== Bernoulli Naive Bayes ===")
print(f"Accuracy: {accuracy_score(y_test, preds_bnb):.4f}")
print(f"F1-Score (Weighted): {f1_score(y_test, preds_bnb, average='weighted'):.4f}\n")
```
