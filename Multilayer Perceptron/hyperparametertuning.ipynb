```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_wine
from sklearn.neural_network import MLPClassifier
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score
```

```python
data = load_wine()
x,y = data.data,data.target
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size = 0.2,random_state = 42)
```

```python
results = []
rates = [0.001,0.01]
sizes = [32,64]
epochs = [150,300]
for rate in rates:
    for size in sizes:
        for epoch in epochs:
            model = MLPClassifier(hidden_layer_sizes = (100,),learning_rate_init = rate,batch_size = size,max_iter = epoch,random_state = 42)
            model.fit(x_train,y_train)
            pred = model.predict(x_test)
            results.append([rate,size,epoch,accuracy_score(y_test,pred)])
```

```python
comparison = pd.DataFrame(results,columns = ['Learning Rate','Batch Size','Epochs','Accuracy'])
comparison
```
