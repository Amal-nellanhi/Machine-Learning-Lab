```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_digits
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from sklearn.preprocessing import StandardScaler
```

```python
data = load_digits()
x = data.data
x = StandardScaler().fit_transform(x)
```

```python
results = []
clusters = [i for i in range(2,11)]
for cluster in clusters:
    model = KMeans(n_clusters = cluster,random_state = 42,n_init = 10)
    pred = model.fit_predict(x)
    results.append([cluster,model.inertia_,silhouette_score(x,pred)])
results
```

```python
comparison = pd.DataFrame(results,columns= ['No of Clusters','Inertia','Silhouette'])
comparison
```
