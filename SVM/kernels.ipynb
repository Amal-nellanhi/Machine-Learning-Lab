```python
import time
from sklearn import metrics
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.datasets import fetch_openml


print("Loading dataset (this may take a moment)...")
X, y = fetch_openml('Fashion-MNIST', version=1, return_X_y=True, as_frame=False, parser='auto')


X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=10000, random_state=42)


n_train, n_test = 4000, 1000
X_train_sub = X_train[:n_train]
y_train_sub = y_train[:n_train]
X_test_sub = X_test[:n_test]
y_test_sub = y_test[:n_test]


scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train_sub)
X_test_scaled = scaler.transform(X_test_sub)


kernels = ['linear', 'poly', 'rbf']
results = {}

for k in kernels:
    
    clf = SVC(kernel=k, random_state=42)
    
   
    start_time = time.time()
    clf.fit(X_train_scaled, y_train_sub)
    train_time = time.time() - start_time
    
    
    y_pred = clf.predict(X_test_scaled)
    acc = metrics.accuracy_score(y_test_sub, y_pred)
    results[k] = {'accuracy': acc, 'train_time': train_time}


print("\nSVM Performance Comparison on Subset (4000 train, 1000 test):")
print("-" * 65)
for k, v in results.items():
    print(f"Kernel: {k.capitalize():<7} | Accuracy: {v['accuracy']*100:.2f}% | Train Time: {v['train_time']:.2f}s")
```
