```python
# Implement Ridge and Lasso regression on the Diabetes dataset. Compare the performance
# of these regularized models with standard linear regression.
# Tasks:

# ● Load and preprocess the dataset.
# ● Implement Ridge and Lasso regression.
# ● Tune hyperparameters using cross-validation.
# ● Compare performance metrics (MSE, R-squared) with standard linear regression.
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_diabetes
from sklearn.linear_model import LinearRegression,RidgeCV,LassoCV
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error,r2_score
```

```python
data = load_diabetes(as_frame = True)
data = data.frame
data
```

```python
x = data.drop('target',axis = 1)
y = data['target']
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size = 0.2,random_state = 42)
```

```python
# Linear regression
linear = LinearRegression()
linear.fit(x_train,y_train)
lin_pred = linear.predict(x_test)
```

```python
# Ridge Regularization
ridge = RidgeCV(alphas = [0.001,0.1,1,10],cv = 5)
ridge.fit(x_train,y_train)
rid_pred = ridge.predict(x_test)
```

```python
# Lasso Regularization
lasso = LassoCV(alphas = [0.001,0.1,1,10], cv = 5)
lasso.fit(x_train,y_train)
las_pred = lasso.predict(x_test)
```

```python
results = []
models = {"Linear Regression":lin_pred,"Ridge":rid_pred,"Lasso":las_pred}
for model,pred in models.items():
    results.append([model,mean_squared_error(y_test,pred),r2_score(y_test,pred)])
```

```python
comparison = pd.DataFrame(results,columns = ["Model","MSE","R2 Score"])
comparison
```

```python
# Best Parameters
print(f"Ridge:{ridge.alpha_}")
print(f"Lasso:{lasso.alpha_}")
```
